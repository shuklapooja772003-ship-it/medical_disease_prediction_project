# Medical Disease Prediction System

**Educational project — not a medical diagnostic tool.**

```
import pandas as pd\nimport matplotlib.pyplot as plt\nimport seaborn as sns\nfrom sklearn.model_selection import train_test_split\nfrom sklearn.ensemble import RandomForestClassifier\nfrom sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```
df=pd.read_csv('../data/medical_symptoms.csv')\ndf.head()
```

```
print(df.shape)\nprint(df.isnull().sum())\ndf['disease'].value_counts()
```

```
df['disease'].value_counts().sort_values().plot(kind='barh',figsize=(8,5)); plt.title('Disease Class Distribution'); plt.show()
```

```
X=df.drop(columns='disease'); y=df['disease']\nX_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
```

```
model=RandomForestClassifier(n_estimators=250,random_state=42,n_jobs=-1)\nmodel.fit(X_train,y_train)\npred=model.predict(X_test)\nprint('Accuracy:',accuracy_score(y_test,pred))\nprint(classification_report(y_test,pred))
```

```
cm=confusion_matrix(y_test,pred,labels=model.classes_)\nsns.heatmap(cm,xticklabels=model.classes_,yticklabels=model.classes_,cmap='Blues')\nplt.xlabel('Predicted'); plt.ylabel('Actual'); plt.xticks(rotation=90); plt.tight_layout(); plt.show()
```

```
sample=pd.DataFrame([[1 if s in ['fever','cough','fatigue','sore_throat'] else 0 for s in X.columns]],columns=X.columns)\nprint('Model output (not a diagnosis):',model.predict(sample)[0])
```
